## Measure of Dispersion

1. Variance

2. Standard Deviation

```python
import numpy as np
import statistics
```

```python
ages = [23, 24, 32, 45, 12, 43, 67, 45, 32, 56, 32, 120]
```

```python
statistics.variance(ages)
```

```python
print(np.var(ages))
```

With statistics.variance variance is calculated as sample data where it is divided by n - 1
and in np.var it is divided by N or It is a population variance

```python
def var(data, dof = 0):
    n = len(data)
    # mean of data
    mean = sum(data)/n
    # variance
    deviation = [(x - mean)**2 for x in data]
    variance = sum(deviation)/(n-dof)
    return variance
```

```python
print(var(ages))
```

```python
import math
std_deviation = math.sqrt(statistics.pvariance(ages))
print(std_deviation)
```
