## Check whether Distribution is Normal Distribution

```python
# If you want to check whether feature is guassian or normal distributed
# Q Q plot (Probability plot)

import matplotlib.pyplot as plt
import scipy.stats as stat
import pylab
```

```python
def plot_data(sample):
    plt.figure(figsize=(10,6))
    plt.subplot(1,2,1)
    sns.histplot(sample)
    plt.subplot(1,2,2)
    stat.probplot(sample, dist='norm', plot=pylab)
    plt.show()
```

```python
# Create a normal distributed data
s = np.random.normal(0.5, 0.2, 1000)
plot_data(s)
```

```python
mu, sigma = 3., 1. # mean and standard deviation
sample = np.random.lognormal(mu, sigma, 100)
```

```python
def plot_log_data(sample):
    plt.figure(figsize=(10,6))
    plt.subplot(1,2,1)
    sns.histplot(sample)
    plt.subplot(1,2,2)
    stat.probplot(np.log(sample), dist='norm', plot=pylab)
    plt.show()
```

```python
plot_log_data(sample)
```

On Q-Q plot:

X-axis (Theoretical quantiles)
→ where points should fall if the data were perfectly normal

Y-axis (Ordered values)
→ your actual data, sorted from smallest to largest

Each dot = expected normal value, actual data value

A Q–Q plot becomes a straight line when your data’s distribution matches the theoretical distribution (here: normal).
